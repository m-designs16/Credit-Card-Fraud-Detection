```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
```

Loading the daatset to panda data frame

```python
credit_card_data=pd.read_csv('creditcard.csv.zip')
```

data set information

checking of missing values in each column

```python
credit_card_data.isnull().sum()
```

```python
credit_card_data['Class'].value_counts()
```

separating the data for analysis

```python
legit = credit_card_data[credit_card_data.Class == 0]
fraud = credit_card_data[credit_card_data.Class ==1]
```

```python
print(legit.shape)
print(fraud.shape)
```

statistical measures of the data

```python
legit.Amount.describe()
```

```python
fraud.Amount.describe()
```

compare the values for both transactions

```python
credit_card_data.groupby('Class').mean()
```

```python
legit_sample = legit.sample(n=492)
```

Concatenating two DataFrames

```python
new_dataset = pd.concat([legit_sample,fraud], axis=0)
```

```python
new_dataset.head()
```

```python
new_dataset.tail()
```

```python
new_dataset['Class'].value_counts()
```

```python
new_dataset.groupby('Class').mean()
```

Splitting the data into  features & Targets

```python
x=new_dataset.drop(columns='Class', axis=1)
y=new_dataset['Class']
```

```python
print(x)
```

```python
print(y)
```

```python
x_train, x_test, y_train, y_test= train_test_split(x, y, test_size=0.2, stratify=y, random_state=2)
```

```python
scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
print(x.shape, x_train.shape, x_test.shape)
```

Model Training

Logistic Regression

```python
model = LogisticRegression(class_weight='balanced')
```

training the Logistic Regression Model with Training Data

```python
model.fit(x_train, y_train)
```

accuracy on training data

```python
x_train_prediction = model.predict(x_train)
training_data_accuracy = accuracy_score(x_train_prediction, y_train)
```

```python
print('Accuracy on Training data : ', training_data_accuracy)
```

accuracy on test data

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

x_test_prediction = model.predict(x_test)

print("Accuracy:", accuracy_score(y_test, x_test_prediction))

print("\nConfusion Matrix:\n", confusion_matrix(y_test, x_test_prediction))

print("\nClassification Report:\n", classification_report(y_test, x_test_prediction))
```

```python
input_data = x_test[0]

# reshape
input_data = input_data.reshape(1, -1)

prediction = model.predict(input_data)

if prediction[0] == 0:
    print("Legitimate Transaction")
else:
    print("Fraudulent Transaction")
```
